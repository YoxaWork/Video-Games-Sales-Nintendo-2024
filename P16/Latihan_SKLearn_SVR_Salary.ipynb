**Tujuan:** menerapkan Support Vector Regression (SVR) dengan SKLearn.

## 1. Ubah data menjadi DataFrame

```python
import pandas as pd

# membaca dataset dan mengubahnya menjadi dataframe
data = pd.read_csv('Salary_Data.csv')
```

Cek missing value. Seharusnya tidak ada.

```python
data.info()
```

```python
data.head()
```

## 2. Pisahkan atribut dan label

Karena hanya ada satu atribut, bentuknya perlu diubah menjadi 2 dimensi agar bisa diterima model SKLearn.

```python
import numpy as np

# memisahkan atribut dan label
X = data['YearsExperience']
y = data['Salary']

# mengubah bentuk atribut
X = np.array(X)
X = X[:,np.newaxis]
```

## 3. Latih model SVR

Parameter: `C = 1000`, `gamma = 0.05`, `kernel = 'rbf'`.

```python
from sklearn.svm import SVR

# membangun model dengan parameter C, gamma, dan kernel
model  = SVR(C=1000, gamma=0.05, kernel='rbf')

# melatih model dengan fungsi fit
model.fit(X,y)
```

## 4. Visualisasi model

```python
import matplotlib.pyplot as plt

# memvisualisasikan model
plt.scatter(X, y)
plt.plot(X, model.predict(X))
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

Garis model belum menyesuaikan pola data dengan baik (skor R² di bawah masih rendah). Untuk memperbaikinya, parameter bisa dicari dengan **grid search** di modul berikutnya.

```python
# skor R2 pada data (tambahan, untuk melihat seberapa pas model)
print('R2:', model.score(X, y))
```
