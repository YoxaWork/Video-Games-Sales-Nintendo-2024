# Latihan SKLearn SVM untuk Klasifikasi

**Dataset:** Pima Indians Diabetes Database (768 sampel, 8 atribut, 1 label `Outcome`)

**Sumber:** https://www.kaggle.com/datasets/uciml/pima-indians-diabetes-database

**Tujuan:** mengklasifikasikan apakah pasien positif diabetes (1) atau tidak (0) menggunakan Support Vector Classifier (SVC).

**Tahapan:**
1. Ubah data ke dalam DataFrame
2. Pisahkan atribut dan label
3. Standarisasi
4. Bagi dataset (train dan test)
5. Buat dan latih model SVC
6. Evaluasi model

> Sebelum menjalankan: download `diabetes.csv` dari link Kaggle di atas, lalu upload ke folder Colab.

## 1. Ubah data ke dalam DataFrame

```python
import pandas as pd

df = pd.read_csv('diabetes.csv')
```

```python
# 5 baris teratas untuk melihat isi dataset
df.head()
```

Cek missing value dan tipe data tiap kolom. Semua kolom seharusnya lengkap dan bertipe numerik (`int64` / `float64`).

```python
df.info()
```

## 2. Pisahkan atribut dan label

```python
# memisahkan atribut pada dataset dan menyimpannya pada sebuah variabel
X = df[df.columns[:8]]

# memisahkan label pada dataset dan menyimpannya pada sebuah variabel
y = df['Outcome']
```

## 3. Standarisasi

Skala tiap kolom berbeda (misalnya `Glucose` dan `DiabetesPedigreeFunction`), jadi nilainya disamakan dengan `StandardScaler`.

```python
from sklearn.preprocessing import StandardScaler

# standarisasi nilai-nilai dari dataset
scaler = StandardScaler()
scaler.fit(X)
X = scaler.transform(X)
```

## 4. Bagi dataset (training dan testing)

```python
from sklearn.model_selection import train_test_split

# memisahkan data untuk training dan testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, random_state=42)

print('Data train:', X_train.shape)
print('Data test :', X_test.shape)
```

## 5. Buat dan latih model SVC

```python
from sklearn.svm import SVC

# membuat objek SVC dan memanggil fungsi fit untuk melatih model
clf = SVC()
clf.fit(X_train, y_train)
```

## 6. Evaluasi model

Skor akurasi prediksi model terhadap data testing.

```python
# Menampilkan skor akurasi prediksi
clf.score(X_test, y_test)
```
